```python
# Remover registros sem peso informado
df_modelagem = df.dropna(subset=["PESO"]).copy()

# Criar variável-alvo
df_modelagem["baixo_peso"] = (
    df_modelagem["PESO"] < 2500
).astype(int)

print("Total de registros válidos:", len(df_modelagem))
print("\nDistribuição da variável-alvo:")
print(df_modelagem["baixo_peso"].value_counts())

print("\nPercentual:")
print(
    df_modelagem["baixo_peso"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
contagem = df_modelagem["baixo_peso"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(
    ["Peso normal/maior", "Baixo peso"],
    contagem.values
)

plt.title("Distribuição de baixo peso ao nascer")
plt.ylabel("Quantidade de registros")
plt.xlabel("Classificação")

plt.show()
```

Nota - se classe minoritária: bebês com baixo peso serão menos numerosos que os demais.

Não devemos avaliar o modelo somente pela acurácia. Por exemplo, um modelo que simplesmente classificasse todos os registros como "sem baixo peso" teria aproximadamente 89,67% de acurácia, mas não identificaria nenhum caso de baixo peso. Por isso vamos considerar principalmente Recall, F2-score e PR-AUC.

Esses resultados indicam associações exploratórias entre faixa etária materna e baixo peso ao nascer, não estabelecendo relações causais.

```python
# Consultas pré-natal(CONSPRENAT)
print("Distribuição das consultas pré-natal:")
print(df_modelagem["CONSPRENAT"].value_counts().sort_index())
```

```python
analise_prenatal = (
    df_modelagem.groupby("CONSPRENAT")["baixo_peso"]
    .agg(["count", "mean"])
)

analise_prenatal["percentual_baixo_peso"] = (
    analise_prenatal["mean"] * 100
)

display(analise_prenatal.round(2))
```

```python
# Agrupar consultas pré-natal
# Criar uma cópia para análise
df_prenatal = df_modelagem.copy()

# Tratar o código 99 como informação ignorada
df_prenatal["consultas_validas"] = (
    df_prenatal["CONSPRENAT"].replace(99, np.nan)
)

# Criar faixas de consultas
df_prenatal["faixa_consultas"] = pd.cut(
    df_prenatal["consultas_validas"],
    bins=[-1, 3, 6, 9, 12, float("inf")],
    labels=[
        "0 a 3",
        "4 a 6",
        "7 a 9",
        "10 a 12",
        "13 ou mais"
    ]
)

# Calcular resultados por faixa
analise_faixas = (
    df_prenatal.groupby(
        "faixa_consultas",
        observed=True
    )["baixo_peso"]
    .agg(["count", "mean"])
)

analise_faixas["percentual_baixo_peso"] = (
    analise_faixas["mean"] * 100
)

display(analise_faixas.round(2))
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    analise_faixas.index.astype(str),
    analise_faixas["percentual_baixo_peso"]
)

plt.title("Baixo peso ao nascer por quantidade de consultas pré-natal")
plt.xlabel("Quantidade de consultas")
plt.ylabel("Percentual de baixo peso (%)")

plt.show()
```

### Análise agrupada das consultas pré-natal

A análise agrupada demonstrou uma associação inversa entre a quantidade de consultas pré-natal e a ocorrência de baixo peso ao nascer.

A faixa de 0 a 3 consultas apresentou a maior proporção de baixo peso (27,18%), seguida pela faixa de 4 a 6 consultas (22,20%). Em contrapartida, a faixa de 10 a 12 consultas apresentou 6,45%, enquanto a faixa de 13 ou mais consultas registrou 6,61%.

Esses resultados indicam que o número de consultas pré-natal é uma variável potencialmente relevante para o modelo preditivo. Entretanto, os resultados são observacionais e podem ser influenciados pela duração da gestação e por outros fatores, não permitindo estabelecer causalidade.

```python
# Mês de inicio do pré-natal (MESPRENAT)
print("Distribuição do mês de início do pré-natal:")

print(
    df_modelagem["MESPRENAT"]
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
# Criar cópia para análise
df_mes = df_modelagem.copy()

# Tratar códigos fora de 1 a 10 como ausentes
df_mes["mes_inicio_valido"] = (
    df_mes["MESPRENAT"]
    .where(df_mes["MESPRENAT"].between(1, 10))
)

analise_mes = (
    df_mes.groupby("mes_inicio_valido")["baixo_peso"]
    .agg(["count", "mean"])
)

analise_mes["percentual_baixo_peso"] = (
    analise_mes["mean"] * 100
)

display(analise_mes.round(2))
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    analise_mes.index.astype(int).astype(str),
    analise_mes["percentual_baixo_peso"]
)

plt.title("Baixo peso por mês de início do pré-natal")
plt.xlabel("Mês de início do pré-natal")
plt.ylabel("Percentual de baixo peso (%)")

plt.show()
```

### Análise do mês de início do pré-natal

A análise da variável MESPRENAT demonstrou diferenças na proporção de baixo peso ao nascer conforme o mês de início do acompanhamento pré-natal.

Os registros com início no primeiro mês apresentaram 9,73% de baixo peso, enquanto aqueles com início no quinto e sexto meses apresentaram 12,92% e 12,70%, respectivamente.

Nos meses finais, os percentuais apresentaram oscilações, porém o número reduzido de observações limita a interpretação dos resultados.

Foram identificados 110 registros com código 99 e 911 valores ausentes, que deverão ser tratados no pré-processamento.

Os resultados indicam uma associação exploratória entre o início do acompanhamento pré-natal e o baixo peso ao nascer, sem estabelecer relação causal.
